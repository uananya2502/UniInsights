# 3 Term Frequency Analysis

## Objective

Term Frequency (TF) analysis was performed to identify the most frequently occurring words in placement-related comments. The objective of this technique is to understand the dominant topics, concerns, and themes discussed by students regarding university placements.

## Why this Technique was Used

Term Frequency analysis provides a simple and effective way to identify commonly discussed topics within a large collection of comments. Frequently occurring words often indicate major areas of interest, concern, or discussion among students.

This technique serves as an initial exploratory analysis before applying advanced methods such as TF-IDF, sentiment analysis, and topic modeling.

## Alternative Approaches

An alternative approach is TF-IDF (Term Frequency–Inverse Document Frequency), which identifies important words rather than simply frequent words.

However, TF analysis was performed first because it provides a direct overview of the most commonly discussed terms within the dataset.

## Why Alternative Was Not Used Initially

TF-IDF may suppress frequently occurring words that are actually important in the context of placement discussions. Therefore, Term Frequency was selected as the first step to understand the overall vocabulary and discussion patterns before applying more advanced weighting techniques.

## Expected Insights

This analysis helps identify commonly discussed placement-related concepts such as internships, packages, recruiters, salary, campus placements, and job opportunities.

# 4 TF-IDF Analysis
#### Objective

TF-IDF (Term Frequency-Inverse Document Frequency) analysis was performed to identify the most important words in placement-related comments.

Unlike Term Frequency, which only measures how often a word appears, TF-IDF assigns greater importance to words that are distinctive and informative within the dataset.

#### Why this Technique was Used

Term Frequency often highlights very common words such as:

placement
student
college
university

which may appear frequently but provide limited analytical insight.

TF-IDF helps identify words that are both:

Frequent

and

Informative

thereby revealing the most significant concepts discussed by students.

#### alternative Approach

Alternative:

Term Frequency (TF)
#### Why TF Alone Was Not Used

TF only measures occurrence frequency.

It cannot distinguish between:

Common words

and

Important words

TF-IDF overcomes this limitation by reducing the importance of words that occur in many comments.

#### Expected Insights

TF-IDF may highlight:

placement

lakh

salary

recruiter

package

lpa

rather than generic words.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

#### Create TF-IDF Matrix

Use the cleaned comments.

```python
tfidf = TfidfVectorizer(

    max_features=1000,

    stop_words='english'

)

tfidf_matrix = tfidf.fit_transform(
    df["clean_comment"]
)
```

#### Calculate Average TF-IDF Score

```python
tfidf_scores = pd.DataFrame({

    "Word":
    tfidf.get_feature_names_out(),

    "TFIDF":
    tfidf_matrix.mean(axis=0).A1

})
```

#### Top Important Words

```python
top_tfidf = (

    tfidf_scores

    .sort_values(
        by="TFIDF",
        ascending=False
    )

    .head(20)

)

top_tfidf
```

```python
df["analysis_text"] = (

    df["tokens"]

    .apply(
        lambda x:
        " ".join(x)
    )

)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=1000
)

tfidf_matrix = tfidf.fit_transform(
    df["analysis_text"]
)
```

```python
tfidf_scores = pd.DataFrame({

    "Word":
    tfidf.get_feature_names_out(),

    "TFIDF":
    tfidf_matrix.mean(axis=0).A1

})

top_tfidf = (

    tfidf_scores

    .sort_values(
        by="TFIDF",
        ascending=False
    )

    .head(30)

)

top_tfidf
```

The analysis revealed that the term "placement" received the highest TF-IDF score, confirming that employment opportunities remain the central focus of student discussions.

Several academic program-related terms such as "CSE", "B.Tech", "MBA", "branch", and "engineering" also appeared among the most important words. This indicates that students often evaluate placement outcomes in relation to their chosen field of study.

Admission-related terms including "admission", "JEE", "rank", and "percentile" were found to be highly significant. This suggests that students frequently associate admission competitiveness and academic pathways with future placement prospects.

Additionally, economic factors such as "fees", "package", and "lakh" appeared prominently, highlighting that students actively compare educational costs with expected career outcomes and salary packages.

Overall, TF-IDF analysis demonstrates that student discussions surrounding placements extend beyond employment opportunities and encompass academic programs, admission processes, and financial considerations.

Objective

Bigram Analysis was performed to identify the most frequently occurring two-word combinations within placement-related comments. Unlike single-word frequency analysis, bigrams preserve contextual information and reveal meaningful relationships between words.

Why this Technique was Used

Single words often lack context.

For example:

does not indicate what aspect of placement is being discussed.

However:

provides much clearer insights.

Bigram analysis helps identify frequently discussed concepts and recurring themes within student discussions.

Alternative Approaches

Alternative:

Term Frequency
Why Not Use Only TF?

Term Frequency analyzes individual words and may lose contextual information.

Bigram analysis preserves word associations and provides more meaningful insights into discussion patterns.

Expected Insights

The analysis is expected to reveal phrases related to:

salary packages

placement opportunities

admission criteria

academic programs

campus recruitment

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
import matplotlib.pyplot as plt
```

```python
bigram_vectorizer = CountVectorizer(

    ngram_range=(2,2),

    max_features=1000

)

X = bigram_vectorizer.fit_transform(
    df["analysis_text"]
)
```

```python
sum_words = X.sum(axis=0)

bigrams = [

    (
        word,
        sum_words[0, idx]
    )

    for word, idx in
    bigram_vectorizer.vocabulary_.items()

]

bigrams = sorted(

    bigrams,

    key=lambda x: x[1],

    reverse=True

)
```

```python
top_bigrams = pd.DataFrame(

    bigrams[:20],

    columns=[
        "Bigram",
        "Frequency"
    ]

)

top_bigrams
```

```python
plt.figure(
    figsize=(12,8)
)

plt.barh(

    top_bigrams["Bigram"],

    top_bigrams["Frequency"]

)

plt.title(
    "Top 20 Bigrams in Placement Comments"
)

plt.xlabel(
    "Frequency"
)

plt.ylabel(
    "Bigrams"
)

plt.gca().invert_yaxis()

save_plot("figure_9.png")
```

```python
custom_stopwords = {

    # Hindi
    "hai","ho","h","ka","ki","ke","ko","se","me","mai","mein",
    "aur","ya","ye","wo","is","us","to","toh","bhi","nhi",
    "nahi","liye","kar","karo","kr","krna","krne","hua","huaa",
    "mujhe","mera","mere","apna","apni","apne","acha","accha",
    "sirf","fir","ab","sab","kuch","bahut","kya","kese","kaise",
    "kon","kaun","kab","kaha","wah","agar",

    # YouTube words
    "video","videos","channel","watch","watched","subscribe",
    "subscribed","like","liked","comment","comments","share",

    # Common conversational words
    "please","pls","plz","sir","madam","bro","bhai","bhaiya",
    "hello","hi","thanks","thank","welcome",

    # Random filler
    "good","best","great","nice","awesome","really","much",
    "also","well","yes","no","ok","okay","get","got","make",
    "made","know","tell","u","ur"

}

stop_words.update(custom_stopwords)
```

```python
def preprocess_topic(text):

    text = str(text).lower()

    # Remove URLs
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)

    # Remove email
    text = re.sub(r"\S+@\S+", "", text)

    # Remove numbers
    text = re.sub(r"\d+", "", text)

    # Remove punctuation
    text = re.sub(r"[^\w\s]", " ", text)

    words = word_tokenize(text)

    words = [

        word

        for word in words

        if len(word) > 2
        and word.isalpha()
        and word not in stop_words

    ]

    return words
```

```python
df["tokens"] = df["comment_text"].apply(preprocess_topic)
```

```python
df = df[df["tokens"].map(len) > 0]
```
