```python
import pandas as pd

video_df = pd.read_csv("india_universities_placement_videos.csv")

video_df.head()
```

```python
print(video_df.columns.tolist())
```

```python
import matplotlib.pyplot as plt

top = video_df["university_name"].value_counts().head(15)

plt.figure(figsize=(12,6))

top.plot(kind="bar")

plt.title("Top 15 Universities by Number of Videos")

plt.xlabel("University")

plt.ylabel("Number of Videos")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
video_df["published_at"] = pd.to_datetime(video_df["published_at"])

video_df["year"] = video_df["published_at"].dt.year

video_df.groupby("year").size().plot(
    kind="line",
    marker="o",
    figsize=(8,5)
)

plt.title("Video Upload Trend")

plt.ylabel("Videos")

plt.show()
```

```python
top_channel = video_df["channel_name"].value_counts().head(15)

top_channel.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Most Active YouTube Channels")

plt.ylabel("Videos")

plt.xticks(rotation=45)

plt.show()
```
